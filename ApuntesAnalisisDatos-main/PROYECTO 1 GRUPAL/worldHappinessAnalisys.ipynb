```python
import kagglehub
import kagglehub

# Descargar la version más reciente
path = kagglehub.dataset_download("agrafintech/world-happiness-index-and-inflation-dataset")
print("Path to dataset files:", path)
```

```python
import pandas as pd
# Leer el archivo
df = pd.read_csv("WHI_Inflation.csv", sep="\t")
df
```

```python
# Eliminación de filas duplicadas
dfLimpio = df.drop_duplicates()
dfLimpio

# Reemplazo de filas con NaN por ceros
dfLimpio = dfLimpio.fillna(0)
dfLimpio
```

```python
# Filtro por pais
dfLimpio[(dfLimpio["Country"] == "Zimbabwe")]
```

```python
# Métricas generales
dfLimpio.describe()
```

```python
# Lista de paises
listaPaises = dfLimpio["Country"].drop_duplicates().reset_index()
listaPaises
```

```python
# Ordenar por Año y Score
felicidadAnualPorPais = dfLimpio[["Country", "Year", "Score"]].sort_values(by=["Year", "Score"], ascending=True)
felicidadAnualPorPais
```

```python
# Top 5 paises más felices
top5PaisesMasFelices = felicidadAnualPorPais.groupby("Country")
top5PaisesMasFelices = top5PaisesMasFelices["Score"].mean().reset_index().sort_values(by="Score", ascending=False)
top5PaisesMasFelices = top5PaisesMasFelices.head()
top5PaisesMasFelices
```

```python
import seaborn as sns

datosTop5PaisesMasFelices = felicidadAnualPorPais[felicidadAnualPorPais["Country"].isin(top5PaisesMasFelices["Country"])]

# Graficar Score VS Year de los 5 paises mas felices
sns.lineplot(data=datosTop5PaisesMasFelices, x="Year", y="Score", hue="Country")
```

```python
import matplotlib.pyplot as plt
# Averiguar la categoría que más afecta a Score mediante matriz de confusión
correlationMatrix = dfLimpio.select_dtypes(include="number").corr() # Selecciona solo las columnas numéricas y saca correlación

plt.figure(figsize=(12, 8))
sns.heatmap(correlationMatrix, annot=True, cmap="seismic", fmt=".2f")
plt.title("Matriz de correlacion")
plt.show()
```
