```python
# Cambiar el NaN  tras haberlo introducido
file["Age"]=file["Age"].fillna(3)
```

```python
# Borrar un valor con NaN
file.dropna()
```

```python
#Agrupar por distintos atributos
agrupacionPorNombre = file.groupby("Name")
horasVistasAcumuladas = agrupacionPorNombre.Watch_Time_Hours.sum()

agrupacionPorSuscripcion = file.groupby("Subscription_Type")
suscritosPorCategoria = agrupacionPorSuscripcion.Name.count()

agrupacionPorPais = file.groupby("Country")
paisesMasCinefilos = agrupacionPorPais.Watch_Time_Hours.sum()

agrupacionPorGenero = file.groupby("Favorite_Genre")
GenerosMasVistos = agrupacionPorGenero.Watch_Time_Hours.sum()

print("horas vistas acumuladas: ", horasVistasAcumuladas)
print("suscriptores por categoría: ", suscritosPorCategoria)
```
