```python
import pandas as pd

# Leer el archivo
file = pd.read_csv("netflix.csv")
```

```python
# 1. Muestra las 10 primeras filas del dataframe. 
copia = file.head(10)
copia
```

```python
# 2. Agrupa los datos por tipo de suscripción y muestra cuántos usuarios están suscritos a cada tipo. 
# (Los resultados de la siguiente imagen se obtienen de las primeras 10 filas del dataframe)  
agrupacionPorSuscripcion = copia.groupby("Subscription_Type")
suscritos = agrupacionPorSuscripcion["Name"].count()
suscritos
```

```python
# 3. Aplica reset_index en el resultado anterior para obtener el siguiente resultado:
suscritos.reset_index()
```

```python
# 4. Muestra el tipo de suscripción que más usuarios tiene.
suscritosOrdenado = suscritos.sort_values(ascending=False)
mayorSuscripcion = suscritosOrdenado.head(1)

mayorSuscripcion
```

```python
# 5. Agrupa el dataframe por país y obtén la persona de mayor edad por cada país, después ordena el resultado de mayor a menor. 
agrupacionPorPais = copia.groupby("Country")
agrupacionPorPaisOrdenada = agrupacionPorPais["Age"].max().sort_values(ascending=False)

agrupacionPorPaisOrdenada
```

```python
# 6. Muestra cuál es el país con la persona de mayor edad. 
paisMasViejo = agrupacionPorPaisOrdenada.head(1)
paisMasViejo
```

```python
# 7. Agrupa el dataframe por género favorito y muestra el total de horas de visualización por cada género, ordenar de mayor a menor.
agrupacionPorGenero = copia.groupby("Favorite_Genre")
agrupacionPorGeneroOrdenada = agrupacionPorGenero["Watch_Time_Hours"].sum().sort_values(ascending=False)

agrupacionPorGeneroOrdenada
```

```python
# 8. Muestra cuál es el género que más se ve. 
generoMasVisto = agrupacionPorGeneroOrdenada.head(1)
generoMasVisto
```

```python
# 9. Muestra el total de horas de visualización de cada género por país y ordena la tabla de mayor a menor. 
agrupacionPorPaisYGenero  = copia.groupby(["Country", "Favorite_Genre"])
horasVistasPorPais=agrupacionPorPaisYGenero.Watch_Time_Hours.sum().sort_values(ascending=False)

horasVistasPorPais.reset_index()
```

```python
# 10. Elimina los países duplicados del resultado anterior para obtener el género que más se ve en cada país y ordena los resultados de mayor a menor. 
generoMasVistoPorPais = horasVistasPorPais.reset_index().drop_duplicates(subset=["Country"], keep='first')
generoMasVistoPorPais
```

```python
# 11. Encuentra el usuario que más ve películas de terror.
agrupacionPorNombreYGenero = file.groupby(["Name", "Favorite_Genre"])
horasVistasPorUsuario = agrupacionPorNombreYGenero.Watch_Time_Hours.sum()
horasVistasPorUsuario = horasVistasPorUsuario.reset_index()

usuariosMasLocos = horasVistasPorUsuario[horasVistasPorUsuario["Favorite_Genre"] == "Horror"].sort_values(by="Watch_Time_Hours", ascending=False)
usuarioMasLoco = usuariosMasLocos.head(1)
usuarioMasLoco
```

```python
# 12. Convierte los datos de la columna Last_login de texto a tipo fecha-hora con el siguiente código:  
file.Last_Login = pd.to_datetime(file.Last_Login)
file

# Ahora agrega las columnas: dia, mes y año al dataframe:
file["dia"] = pd.DatetimeIndex(file.Last_Login).day
file["mes"] = pd.DatetimeIndex(file.Last_Login).month
file["año"] = pd.DatetimeIndex(file.Last_Login).year

file
```

```python
# 13. Muestra cuál es el género más visto en cada mes sin importar el año.
agrupacionPorGeneroYMes = file.groupby(["Favorite_Genre", "mes"])
horasDeGeneroPorMes = agrupacionPorGeneroYMes.Watch_Time_Hours.sum()
horasDeGeneroPorMes.reset_index().sort_values(by=["mes", "Watch_Time_Hours"],ascending=[True, False]).drop_duplicates(subset=["mes"], keep='first')
```

```python
# 14. Muestra el género más visto en cada año
agrupacionPorAñoYGenero = file.groupby(["año", "Favorite_Genre"])
horasDeGeneroPorAño = agrupacionPorAñoYGenero.Watch_Time_Hours.sum().sort_values(ascending=False)

GeneroMasVistoPorAño = horasDeGeneroPorAño.groupby("año").head(1)
GeneroMasVistoPorAño.reset_index()
```

```python
# 15. Haga un top 3 de los géneros que más ven los menores de edad en el mes de diciembre sin importar el año.
agrupacionPorGeneroYEdad = file.groupby(["Favorite_Genre", "Age", "mes"])
horasVistasPorEdad = agrupacionPorGeneroYEdad.Watch_Time_Hours.sum()

#Filtro por menores de edad
horasVistasPorEdad = horasVistasPorEdad.reset_index()
horasVistasPorNiños = horasVistasPorEdad[ (horasVistasPorEdad["Age"] < 18) & (horasVistasPorEdad["mes"] == 12) ]

# Filtro por género
horasVistasPorGenero = horasVistasPorNiños.groupby("Favorite_Genre")
top3GenerosMasVistosPorNiñosEnDiciembre = horasVistasPorGenero.Watch_Time_Hours.sum().sort_values(ascending=False).head(3)
top3GenerosMasVistosPorNiñosEnDiciembre.reset_index()
```
