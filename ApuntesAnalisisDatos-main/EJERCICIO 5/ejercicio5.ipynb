```python
# 1. Usando el dataset de pingüinos, calcula el promedio del largo del pico:

# Eliminar duplicados
pinguinosLimpio = pinguinos.drop_duplicates()

# Eliminar NaN
pinguinosLimpio = pinguinosLimpio.dropna(axis=0)

# Agrupar por especie de pingüino
pinguinosPorEspecie = pinguinosLimpio.groupby("species")

# Promedio del largo de pico por especie
largoPicoPromedio = pinguinosPorEspecie["bill_length_mm"].mean()
largoPicoPromedio.reset_index()
```

```python
# Grafico de los promedios con Pandas
largoPicoPromedio.plot.bar(x="species", y="bill_length_mm")
```

```python
# Grafico de los promedios con Matplotlib
largoPicoPromedio = largoPicoPromedio.reset_index()
plt.bar(largoPicoPromedio["species"], largoPicoPromedio["bill_length_mm"])
plt.show()
```

```python
# Gráfico de los promedios con Seaborn
sns.barplot(largoPicoPromedio, x="species", y="bill_length_mm")
```

```python
# Gráfico de los promedios usando Plotly
grafico = px.bar(largoPicoPromedio, x="species", y="bill_length_mm")
grafico.show()
```

```python
# 2. Usando el dataset de vuelos, calcula el total de pasajeros por año (La tabla de ejemplo contiene solamente las primeras 5 filas)
vuelos
```

```python
# Eliminar duplicados y NaN
vuelosLimpio = vuelos.drop_duplicates()
vuelosLimpio = vuelosLimpio.dropna(axis=0)
```

```python
# 6. Realiza un gráfico circular o de torta, utilizando el dataset del Titanic, hallando el número de pasajeros por clase. Utiliza las 4 librerías de Python. 
titanic
```

```python
# Hallar numero de pasajeros por clase ( 1) Limpiar de clones y NaNs, 2) agrupar por clase, 3) contar la categoria pclass )
# Limpiar clones
titanicLimpio = titanic.drop_duplicates(keep="first")
titanicLimpio

# Agrupar por clase
titanicPorClase = titanicLimpio.groupby("class")

# Personas en cada clase
personasPorClase = titanicPorClase["pclass"].count()
personasPorClase = personasPorClase.reset_index()
personasPorClase = personasPorClase.rename(columns={"pclass": "people"})
personasPorClase




```

```python
# titanicLimpio = titanicLimpio.dropna()
# titanicLimpio
```

```python
#Gráfico de torta con Pandas
personasPorClase.set_index("class")["people"].plot(kind="pie", autopct="%1.1f%%")

# Personalizar el gráfico
plt.title("Gráfico de torta con Pandas")
plt.show()
```

```python
# Gráfico de torta con Matplotlib
plt.pie(personasPorClase["people"], labels=personasPorClase["class"], autopct="%1.1f%%")
# Personalizar el gráfico
plt.title("Gráfico de torta con Matplotlib")
plt.show()

```

```python
# Gráfico de torta con Seaborn
colors = sns.color_palette("pastel")
plt.pie(personasPorClase["people"], labels=personasPorClase["class"], autopct='%1.1f%%', colors=colors)


# Personalizar el gráfico
plt.title("Gráfico de torta con Pandas")
plt.show()
```

```python
# Gráfico de torta con Plotly
grafico = px.pie(personasPorClase, 
                 names=personasPorClase["class"],
                 values=personasPorClase["people"],
                 title="Gráfico de torta con Plotly")

grafico.update_layout(margin=dict(l=5, r=5, t=30, b=10))
grafico.update_traces(direction="counterclockwise", rotation=90)
grafico.show()
```
